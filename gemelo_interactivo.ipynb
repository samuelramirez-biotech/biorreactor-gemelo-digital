```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
%matplotlib inline

def simular_biorreactor(mu_max=0.40, Ks=0.18, S0=20.0, X0=0.10):
    Yxs = 0.50

    def modelo(t, y):
        X, S = y
        if S < 0: S = 0
        mu = mu_max * S / (Ks + S)
        return [mu * X, -(mu / Yxs) * X]

    sol = solve_ivp(modelo, [0, 30], [X0, S0], t_eval=np.linspace(0, 30, 500))

    plt.close('all')  # <- esto borra la gráfica anterior antes de dibujar la nueva
    fig, ax = plt.subplots(figsize=(10, 4))
    ax.plot(sol.t, sol.y[0], 'g-', linewidth=2, label='Biomasa (g/L)')
    ax.plot(sol.t, sol.y[1], color='orange', linewidth=2, label='Glucosa (g/L)')
    ax.set_xlabel('Tiempo (horas)')
    ax.set_ylabel('Concentración (g/L)')
    ax.set_title(f'μ_max={mu_max} | Ks={Ks} | S₀={S0} g/L | X₀={X0} g/L')
    ax.legend()
    ax.grid(True)
    plt.show()
    print(f"Biomasa final: {round(sol.y[0][-1], 2)} g/L")

interact(
    simular_biorreactor,
    mu_max = FloatSlider(min=0.10, max=0.80, step=0.05, value=0.40, description='μ_max'),
    Ks     = FloatSlider(min=0.01, max=1.00, step=0.05, value=0.18, description='Ks'),
    S0     = FloatSlider(min=5.0,  max=50.0, step=1.0,  value=20.0, description='S₀ inicial'),
    X0     = FloatSlider(min=0.01, max=2.00, step=0.05, value=0.10, description='X₀ inicial')
)
```
